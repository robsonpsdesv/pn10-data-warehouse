```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from treinar_ensemble import carregar_dados_dw, avaliar_modelos

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
%matplotlib inline
```

## 1. Carregamento dos Dados do Data Warehouse

```python
df = carregar_dados_dw()
print(f"Dimensões do dataset: {df.shape}")
df.head()
```

## 2. Análise Exploratória de Risco de Cancelamento

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df, x='indicador_cancelado', y='dias_para_primeiro_orcamento', ax=axes[0], palette='Set2')
axes[0].set_title('Dias até o 1º Orçamento vs Cancelamento', fontweight='bold')
axes[0].set_xticklabels(['Concluído / Ativo', 'Cancelado'])

sns.countplot(data=df, x='qtd_orcamentos_recebidos', hue='indicador_cancelado', ax=axes[1], palette='Set1')
axes[1].set_title('Qtd de Orçamentos vs Cancelamento', fontweight='bold')
axes[1].legend(['Concluído', 'Cancelado'])

plt.tight_layout()
plt.show()
```
